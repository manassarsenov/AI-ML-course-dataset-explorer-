## 1) 🌸 Load the data

We use the built-in **Iris dataset**:
- **150 flowers**
- **4 measurements** for each flower
- **3 species**

**Task:** use the four measurements to predict the flower species.

```python
# Core tools used in this small demo
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display
from sklearn.datasets import load_iris
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split

# Load the built-in Iris dataset
iris = load_iris()

print("✅ Dataset loaded")
print(f"Samples: {iris.data.shape[0]}")
print(f"Measurements per flower: {iris.data.shape[1]}")
print(f"Species: {list(iris.target_names)}")
```

---
## 2) 👀 Look at the data

Before training a model, we inspect the table.

- Each **row** is one flower.
- The first four columns are the **input measurements**.
- `species` is the **answer / target**.

```python
# Create a readable table
flower_table = pd.DataFrame(iris.data, columns=iris.feature_names)
flower_table["species"] = [iris.target_names[index] for index in iris.target]

print("✅ First 10 rows")
display(flower_table.head(10))

print("✅ Number of flowers in each species")
display(flower_table["species"].value_counts().rename_axis("species").to_frame("count"))
```

### 🧠 Important lesson

- **X** = the input features: four flower measurements
- **y** = the target: flower species

For a new flower, we know the measurements. We ask the model to predict the species.

---
## 3) ✂️ Split into training and testing sets

We keep part of the data separate.

- **Training set**: used by the model to learn patterns
- **Test set**: used later to check the trained model on unseen rows

**Training and testing are different steps.**

```python
# X contains only the four measurements
X = iris.data

# y contains the known species labels
y = iris.target
```

```python
# Keep 20% of the rows for testing
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
)

print("✅ Data split complete")
print(f"Training rows: {len(X_train)}")
print(f"Testing rows:  {len(X_test)}")
```

---
## 4) 🌲 Train the model

We use a **Random Forest Classifier**.

You do not need the algorithm details today. Notice only this:

`model.fit(X_train, y_train)` = **the model learns from known examples**.

```python
# Create the model
model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
)

# Train the model
model.fit(X_train, y_train)

print("✅ Training complete")
```

---
## 5) 🔮 Make predictions

Now the trained model receives test measurements and returns predicted species.

`model.predict(X_test)` = **use the learned pattern on new rows**.

```python
# Predict the species for every row in the test set
predictions = model.predict(X_test)

print("✅ Predictions created")
print(f"Number of predictions: {len(predictions)}")
```

```python
# Compare actual and predicted labels for the first 10 test rows
results = pd.DataFrame({
    "Actual": [iris.target_names[index] for index in y_test[:10]],
    "Predicted": [iris.target_names[index] for index in predictions[:10]],
})
results["Correct?"] = results["Actual"] == results["Predicted"]

display(results)
```

### One actual prediction

The next cell shows:
1. the four input measurements
2. the actual species
3. the model prediction

```python
# Use the first row from the test set as one clear example
example_number = 0
example_measurements = X_test[example_number]
example_actual = iris.target_names[y_test[example_number]]
example_prediction = iris.target_names[predictions[example_number]]

example_table = pd.DataFrame(
    [example_measurements],
    columns=iris.feature_names,
)
example_table["actual_species"] = example_actual
example_table["predicted_species"] = example_prediction

print("✅ One test example")
display(example_table)
```

---
## 6) 📊 Check accuracy

Accuracy answers one simple question:

**How many test predictions were correct?**

This is one small test split. A high score here does **not** prove that a model is perfect in every situation.

```python
accuracy = accuracy_score(y_test, predictions)
correct_predictions = int((predictions == y_test).sum())

print(f"Accuracy: {accuracy:.1%}")
print(f"Correct predictions: {correct_predictions}/{len(y_test)}")
```

---
## 7) 📈 Quick visualization

We plot petal length and petal width.

The groups are not perfectly identical, but the visible patterns help explain why a model can learn to separate the species.

```python
fig, ax = plt.subplots(figsize=(8, 5))

for species_index, species_name in enumerate(iris.target_names):
    mask = iris.target == species_index
    ax.scatter(
        iris.data[mask, 2],
        iris.data[mask, 3],
        label=species_name,
        alpha=0.75,
    )

ax.set_xlabel("Petal length (cm)")
ax.set_ylabel("Petal width (cm)")
ax.set_title("Iris species by petal dimensions")
ax.legend()
plt.tight_layout()
plt.show()
```
