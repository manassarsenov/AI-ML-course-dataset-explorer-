## C09 — Calculate five scores with NumPy
**Predict:** Five rows and three weights should give **one score per row**.

`@` here does the same multiply-and-add operation for every row.

```python
matrix = five.to_numpy(dtype=float)
weights = np.array([0.5, 0.3, -0.2])
scores = matrix @ weights

print('Shapes:', matrix.shape, '@', weights.shape, '→', scores.shape)
for i, score in enumerate(scores, start=1):
    print(f'Product {i} score: {score:.4f}')
```

## C13 — Order can be wrong even if code runs

`[::-1]` simply reads the array backwards.

```python
print('Original:', a @ w)
print('Reverse values only:', a[::-1] @ w)
print('Reverse both:', a[::-1] @ w[::-1])

first_row = matrix[0]
print('Real Product 1, original:', first_row @ weights)
print('Real Product 1, values reversed only:', first_row[::-1] @ weights)
print('Real Product 1, both reversed:', first_row[::-1] @ weights[::-1])
```
