## 02 — Values and their types (Slide 3)
A variable gives a value a name. `type()` tells you what kind of value it is.

**Look:** Number `80` and text `"80"` are not the same kind of value.

```python
age = 22                 # int: whole number
gpa = 3.85               # float: decimal number
name = "Aziz"            # str: text
is_student = True        # bool: True or False
scores = [80, 90, 70]    # list: many values

for value in [age, gpa, name, is_student, scores]:
    print(value, type(value))

print(80 + 10)
print("80" + "10")
```

## 03 — List, dictionary, loop (Slide 4)
A list keeps several scores together. A dictionary names values with keys. A loop repeats an action; `enumerate()` also gives the index.

```python
scores = [80, 90, 70]
record = {"name": "Aziz", "score": 80}
print("Dictionary score:", record["score"])

for position, score in enumerate(scores):
    print("Index", position, "score", score)
```

## 04 — First average function (Slide 5)
A function takes input and returns a result. **This first version has no protection for an empty list.** Keep it unchanged until the AI-fix step.

```python
def calculate_average(numbers):
    total = sum(numbers)
    count = len(numbers)
    return total / count
```

```python
print("Normal input:", calculate_average([80, 90, 70]))
```
